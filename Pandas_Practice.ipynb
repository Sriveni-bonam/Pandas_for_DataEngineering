**loc (Label-based):**
*   Uses the explicit names of rows and columns (e.g., df.loc['row_name', 'col_name'])
*   Includes both start and end points when slicing.

**.iloc (Integer position-based):**
*   Uses whole numbers for row and column order (e.g., df.iloc[0, 2])
*   Follows standard Python behavior by including the start index and excluding the end index when slicing.
*   Returns an error if you try to use a number that is out of bounds.

```python
import json
file=open('orders_ETL.json')
data=json.load(file)
```

```python
data
```

```python
data[0]
```

*   orders - orders_id, order_date, total_amount, customer_id, product_id, quantity
*   customers - customer_id, cust_name, email, address
*   products - product_id, prod_name, category, price



```python
order_id=[]  # order_table
order_date=[]
total_amount=[]
customer_id=[]
product_id=[]
quantity=[]
cust_id=[] # customer_table
cust_name=[]
email=[]
address=[]
prod_name=[] # product_table
category=[]
price=[]
```

```python
for row in data:
  cust_id.append(row['customer']['customer_id'])
  cust_name.append(row['customer']['name'])
  email.append(row['customer']['email'])
  address.append(row['customer']['address'])
  for product in row['products']:
    order_id.append(row['order_id'])
    order_date.append(row['order_date'])
    total_amount.append(row['total_amount'])
    quantity.append(product['quantity'])
    customer_id.append(row['customer']['customer_id'])
    product_id.append(product['product_id'])
    prod_name.append(product['name'])
    category.append(product['category'])
    price.append(product['price'])
```

```python

order_df=pd.DataFrame({
    'order_id': order_id,
    'order_date': order_date,
    'total_amount': total_amount,
    'quantity': quantity,
    'customer_id': customer_id,
    'product_id': product_id
})
```

```python
order_df
```

```python
for cust_data in data:
    cust_id.append(cust_data['customer']['customer_id'])
    cust_name.append(cust_data['customer']['name'])
    email.append(cust_data['customer']['email'])
    address.append(cust_data['customer']['address'])
```

```python
customer_df=pd.DataFrame({
    'customer_id': cust_id,
    'name': cust_name,
    'email':email,
    'address': address
})
```

```python
customer_df
```

```python
product_df=pd.DataFrame({
    'product_id': product_id,
    'name': prod_name,
    'category':category,
    'price': price
})
```

```python
product_df
```

```python
customer_df.drop_duplicates(subset='customer_id',keep='last',inplace=True)
```

```python
customer_df
```

```python
product_df.drop_duplicates(subset='product_id',keep='last',inplace=True)
product_df
```

```python
order_df.to_sql('orders', engine, index=False, if_exists='replace')
customer_df.to_sql('customers', engine, index=False, if_exists='replace')
product_df.to_sql('products', engine, index=False, if_exists='replace')
```
